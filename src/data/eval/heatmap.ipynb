```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
```

# XGBoost RF on Shape Test Data

```python
df = pd.read_csv('evaluated_shape_test.csv_xgboost_rf.csv')
```

```python
y_true = df['label']
y_pred = df['predictions']

cm = confusion_matrix(y_true, y_pred)
cm_normalized = cm.astype('float') / cm.sum(axis=1)[:, np.newaxis]
```

```python
# Create a labeled heatmap
labels = ['Negative', 'Positive']  # or [0, 1] depending on preference
sns.set(style='whitegrid')

plt.figure(figsize=(6, 5))
sns.heatmap(cm_normalized, annot=True, fmt='.2f', cmap='YlGnBu',
            xticklabels=labels, yticklabels=labels)

plt.xlabel('Predicted')
plt.ylabel('True')
plt.title('Confusion Matrix Heatmap (XGBoost RF on Shape Data)')
plt.tight_layout()
plt.show()
```
